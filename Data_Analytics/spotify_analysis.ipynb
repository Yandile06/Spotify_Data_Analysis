```python
import pandas as pd

df=pd.read_csv("../Data/spotify_data_clean.csv")

df.info()
```

```python
df=df.dropna(subset=['artist_name'])
```

```python
df['artist_genres']=df['artist_genres'].fillna("Unknown")
df.info()
```

```python
text_cols= ['track_id','track_name','artist_name','album_name','album_id','album_type']
for col in text_cols:
    df[col]=df[col].str.strip()
```

```python

df['release_year']=pd.to_datetime(df['album_release_date'],errors='coerce').dt.year # Since thw year is a string value it will be difficult to Manipulate. we need it as quantity
df['release_year']=df['release_year'].astype('Int64')
df.info()
```

```python
# Analysis On The Data 
#Quetion 1: Albums  vs  Sinlges: Popularity Contest.
album_vs_singles=df.groupby('album_type')['track_popularity'].mean().reset_index()
album_vs_singles=album_vs_singles.sort_values(by='track_popularity',ascending=False)
print(album_vs_singles)
```

```python
# For Plotting we are Using SeaBorn
import seaborn as sns
import matplotlib.pyplot as plt

#Creating a barplot  comparing popularity across album types
sns.barplot(data=df ,x='album_type', y='track_popularity', ci=None)
plt.title('Average Track Popularity: Album vs Singles')
plt.xlabel('Album Type')
plt.ylabel('Average Popularity')
plt.show()
```
